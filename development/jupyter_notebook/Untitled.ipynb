```python
from pyspark.sql import SparkSession

# Si Jupyter ya creó Spark, detenerlo
try:
    spark.stop()
except:
    pass

spark = (
    SparkSession.builder
    .master("local[*]")
    .appName("Iceberg Local")

    .config(
        "spark.sql.extensions",
        "org.apache.iceberg.spark.extensions.IcebergSparkSessionExtensions"
    )

    .config(
        "spark.sql.catalog.local",
        "org.apache.iceberg.spark.SparkCatalog"
    )

    .config(
        "spark.sql.catalog.local.type",
        "hadoop"
    )

    .config(
        "spark.sql.catalog.local.warehouse",
        "/home/jovyan/warehouse"
    )

    .getOrCreate()
)
```

```python
spark.conf.get("spark.sql.catalog.local")
```

```python
spark.sql("""
CREATE NAMESPACE IF NOT EXISTS local.fes
""")
```

```python
spark.sql("""
CREATE TABLE IF NOT EXISTS local.fes.alumnos (
    id BIGINT,
    nombre STRING,
    carrera STRING,
    semestre INT
)
USING iceberg
""")
```

```python
spark.sql("""
INSERT INTO local.fes.alumnos
VALUES
    (1, 'Josué', 'MAC', 8),
    (2, 'Juan', 'MAC', 6),
    (3, 'María', 'MAC', 4)
""")
```

```python
spark.sql("""
SELECT *
FROM local.fes.alumnos
""").show()
```

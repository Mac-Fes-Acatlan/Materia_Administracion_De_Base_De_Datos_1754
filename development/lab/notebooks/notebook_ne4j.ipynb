```python
from neo4j import GraphDatabase

URI = "bolt://172.18.0.4:7687"
AUTH = ("neo4j", "k6M9pXw2")
```

```python
df = pd.DataFrame({
    "cliente":  ["Ana", "Ana", "Luis", "María"],
    "producto": ["Laptop", "Mouse", "Laptop", "Teclado"],
    "monto":    [1200, 25, 1150, 80],
})
```

```python
df
```

```python
# 3. Crear nodos y relaciones en una sola consulta
with GraphDatabase.driver(URI, auth=AUTH) as driver:
    driver.execute_query("""
        UNWIND $filas AS f
        MERGE (c:Cliente {nombre: f.cliente})
        MERGE (p:Producto {nombre: f.producto})
        MERGE (c)-[r:COMPRO]->(p)
        SET r.monto = f.monto
    """, filas=df.to_dict("records"))

print("Listo")
```

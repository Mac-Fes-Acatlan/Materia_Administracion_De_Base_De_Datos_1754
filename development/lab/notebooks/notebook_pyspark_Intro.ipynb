```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("engine").getOrCreate()

jdbc_url = "jdbc:postgresql://postgres:5432/db_acatlan"
props = {
    "user": "acatlan",
    "password": "8g4izJPX",
    "driver": "org.postgresql.Driver",
}
```

# Extracción 

```python
df = (
    spark.read
    .format("jdbc")
    .option("url", jdbc_url)
    .option("dbtable", "public.orders")
    .option("user", props["user"])
    .option("password", props["password"])
    .option("driver", props["driver"])
    .load()
)

df.show(2,vertical=True)
```

# Transformación

```python
df = df.filter(df["order_status"] != "delivered") 
```

```python
df.show(2,vertical=True)
```

# Carga

### Parquet

```python
(
    df.write
    .option("header","true")
    .mode("overwrite")
    .parquet("/home/jovyan/notebooks/data_parquet")
)
```

```python
(
    df.repartition(20).write
    .option("header","true")
    .mode("overwrite")
    .parquet("/home/jovyan/notebooks/data_parquet")
)
```

```python
(
    df.write
    .option("header","true")
    .mode("overwrite")
    .partitionBy("order_status")
    .parquet("/home/jovyan/notebooks/data_parquet")
)
```

### CSV

```python
(
    df.write
    .option("header","true")
    .mode("overwrite")
    .csv("/home/jovyan/notebooks/data_csv")

)
```
